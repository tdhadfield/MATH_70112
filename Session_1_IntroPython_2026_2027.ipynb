### Interpreted vs. Compiled programming languages

**Interpreted :** code is executed as you write it without any internal optimisation. 
One of the benefits is that it is easy to and quick to implement, at the cost of slower execution time.<br>
*Examples:* PHP, Ruby, Python, JavaScript

**Compiled :** As the name suggests, there is an intermediary step between writting your code and actually executing it. This step called compilation. During compilation, the code is optimised for the particular hardware and processor available and substantially improves execution time compared to interpreted languages.<br>
*Examples:* C, C++, Erlang, Haskell, Rust, Go

### Copying lists.....
Since a `list` is passed by reference, if you make a assign copy and modify it, the original `list` will also change.

```python
my_list = [1, 2, 3]

my_copy = my_list

my_copy[0] = 10

print(my_copy)

print(my_list)
```

To avoid this issue we need to make a `deepcopy` using the copy module:

```python
import copy

my_list = [1, 2, 3]

my_copy = copy.deepcopy(my_list)

my_copy[0] = 10

print(my_copy)

print(my_list)
```

### `list` comprehension 
A `list` allows to run a `for` loop in a single line and output the result into a `list`.

```python
my_list = [1, 2, 3, 4, 5]

my_list_squared = [element**2 for element in my_list]

print(my_list_squared)
```

The same result is obtained using the (longer) piece of code

```python
my_list = [1, 2, 3, 4, 5]
my_list_squared = [] # initialise empty list

for element in my_list: 
    my_list_squared.append(element**2)
    
print(my_list_squared)
```

Are `list` comprehensions faster than loops?
And what about readability?

```python
import time
import random
N = 10_000_000

my_list = [random.randint(1, 10) for _ in range(N)]
```

```python
t0 = time.time()
my_list_squared = [element**2 for element in my_list]
t_list = time.time() - t0

print("Computation time for the list comprehension: ", t_list)
```

```python
t0 = time.time()

my_list_squared = []
for element in my_list: 
    my_list_squared.append(element**2)
    
t_loop = time.time() - t0
print("Computation time for the list comprehension: ", t_loop)
```

The `list` comprehension is about twice as fast as a `for` loop.

## Dictionaries

```python
print(stock_dict['AAPL'])

print("AAPL open is :", stock_dict['AAPL']["open"])

print("symbols in my dictionary are:", stock_dict.keys())

for symbol in stock_dict.keys():
    print(symbol, "high of the day is ", stock_dict[symbol]["high"])
```

Let us check the following formula, valid for any integer $n$:
$$
\sum_{i=1}^{n} i  = \frac{n(n+1)}{2}
$$

```python
def sumInteg(n):
    res = 0
    for i in range(1, n+1): 
        res += i
    return res

###Using List comprehension we can write the above function in a single line
def sumInteg_list_comprehension(n):
    return sum([i for i in range(1, n+1)])

#Here goes the theoretical formula
def sumInteg_Formula(n):
    return n*(n+1) / 2
```

```python
n = 1000
print("Sum:", sumInteg(n))
print("Sum2:", sumInteg_list_comprehension(n))
print("Theoretical:", sumInteg_Formula(n))
```

## Execution time
While a code has to work and be written clearly, it also has to run fast.
We may therefore need to compute the execution time of a function or of a series of operations that can be written in multiple manners.
The `magic` command`%timeit` provides this functionality.

```python
%timeit sum(range(1000))
```

Since this operation is very fast, `%timeit` automatically does a large number of repetitions. For slower commands, it may adjust the number of repetitions:

```python
%timeit sum(range(1_000_000))
```

The `%time` magic command can be equivalently useful to avoid many repetitions of the same computation.

```python
%time sum(range(100000))
```

'Wall time' corresponds to the total amount of time from start to finish of the computation.

'CPU time' is a measure of parallel efficiency of the algorithm.
It ranges same as 'Wall time' (in the case of serial processing) to  larger than the latter.

## Timing

```python
def sleeper():
    time.sleep(1.75)

def spinlock():
    for _ in range(100_000_000):
        pass

for function in sleeper, spinlock:
    t1 = time.perf_counter(), time.process_time()
    function()
    t2 = time.perf_counter(), time.process_time()
    print(f"{function.__name__}()")
    print(f" Real time: {t2[0] - t1[0]:.2f} seconds")
    print(f" CPU time: {t2[1] - t1[1]:.2f} seconds")
    print()
```

- `time.perf_counter()`: elapsed real time
- `time.process_time()`: CPU time

```python
import numpy as np
import time
```

```python
def f(i,j):
    return i/j
```

```python
def loop_sum(range_m, range_n):
    res = 0
    for i in range_m:
        for j in range_n:
           res += f(i,j)
    return res
```

#### The `itertools` package
`itertools` consists of a set of functions to create iterators for efficient looping.

https://docs.python.org/3/library/itertools.html

In particular, `itertools.product` returns the Cartesian product of input iterables:
for any two sets $A$ and $B$,
$$
A\times B = \{(a,b): a\in A, b\in B\}.
$$

```python
from itertools import product
```

```python
def itertools_sum(range_m, range_n):
    res = 0
    for (i,j) in product(range_m, range_n):
        res += f(i,j)
    return res
```

```python
print("Iter result: ", itertools_sum(range_m,range_n))
```

```python
m, n = 100, 10000
range_m = range(1, m+1)
range_n = range(1, n+1)

t0 = time.time()
loop_sum_value = loop_sum(range_m, range_n)
dt_loop = time.time() - t0

t0 = time.time()
iter_loop_value = itertools_sum(range_m, range_n)
dt_iter = time.time() - t0

print("value:", np.round(loop_sum_value, 2), "|| time:", dt_loop)
print("value:", np.round(iter_loop_value, 2), "|| time:", dt_iter)
```

## logging

Python’s built-in library includes the `logging` module, 
which allows to:

- write messages at different severity levels

- Write log output to the console, files, ..., 

- ...

The three standard levels of log records are the following:

```python
import logging

logging.warning("WARNING: Some data are missing")

logging.error("ERROR: Division by zero")

logging.critical("CRITICAL: Exam failed!!")
```

Equivalently:

```python
logging.log(logging.ERROR, "ERROR: Division by zero")
```

There are in fact two lower-level records, `debug` and `info`, but which do not output anything by default.

Log records can also be accessed by level of severity:

```python
# Log at a level determined at runtime

# Inspect the numeric value of each level
for name in ("DEBUG", "INFO", "WARNING", "ERROR", "CRITICAL"):
    print("Level=", logging.getLevelName(name), ":", name)
    #print(f"{name:8s} = {logging.getLevelName(name)}")
```

### `logging` setup and example

```python
import logging
import time
import sys

log_file = "trade_log.log"

logging.basicConfig(
    level = logging.INFO,
    format = "%(asctime)s  %(levelname)-8s  %(message)s",
    filename = log_file,
    filemode = "w",
    force = True
)

logging.info("Trading app started")
time.sleep(1)
logging.info("Connected to data provider")
logging.warning("Buying order received")
logging.error("Buying order rejected: insufficient funds")

# Read back the file to verify
with open(log_file) as f:
    print(f.read())
```
